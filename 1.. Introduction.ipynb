## 3.. Loop Engineering: 10 Prompts vs 1 Loop

```python
'''


Lets have a quick understanding of How Loop engineering actually changed the way we talk to AI !


Imagine u are a S/W engineer at your company & Your manager asks you to build a Food Delivery App where User can : 
    Browse dishes like Biryani or Pizza.
    Add items to the cart.
    Pay using UPI, Credit Card, or Net Banking.
    Finally, reach the Order Confirmation screen



User ->> AI
You give a prompt to an CHATGPT / CLAUDE and say "Build a Food Delivery App with these features"

The AI generates an output...
And that answer is immediately returned to you.
That's it.

But when you run the application.
    Maybe the payment flow doesn't work..
    Maybe the order confirmation page has bugs..
    

But why u have these mistakes ,  bcz **the AI never checks its own work.** 
So It's your responsibility to explain the issue by rewriting the prompt and try again & test it !






Now using Loop :

    Instead of giving prompt to AI , we design a Loop in such a way that : 

    The AI first generates an output...
    Then verifies whether the answer is correct or not...
    Takes feedback...
    Improves the response...
    And keeps repeating this cycle until the required quality is achieved.

    That's the core idea behind **Loop Engineering**.




In short , So instead of the human deciding what to do next...
    Agent will plan, decide & act & decide whether result is good enough, 
    if result is not good enough ,  Agent will again go into loop & do again !

    So , 
    Your job is no longer to write dozens of prompts.
    Your job is to design the loop, while the AI Agent takes care of the repetitive work.


And throughout this course, you'll learn exactly how to build these intelligent loops yourself.





'''
```

```python
'''

Day-2-day problem :

    Studying for an Exam 📚
    You solve a practice question.

    Repeat:

    Solve a question.
    Check the answer.
    If it's wrong, learn the concept and try another.

    Stop: When you've completed all questions or mastered the topic.




Imagine you're solving practice questions.
You solve the first question, then check your answer.

If it's wrong, you learn the concept, fix your mistake, and solve the next question.
Again, you check the answer.

You keep repeating this process—solve → check → improve—
until you've either completed all the questions or you're confident you've mastered the topic.
That's a loop.


You're repeating the same set of steps until you reach your goal. 
AI loops work exactly the same way. 
Instead of a student solving questions, the AI performs a task, checks its work, fixes mistakes, 
and repeats until the job is done.

Thats a foundation of Loop engineering !!


'''
```

```python
'''



    Now let me give u quick Idea how u can build your first Loop !
    Imagine u are a content creator/Youtuber & u upload videos on Daily basis !
    
    
    To do that, we use ChatGPT/CLAUDE normally, the workflow looks like this:
    Every day you write:

    Search today's AI news.
    Write a script.
    Generate a title.
    Create a thumbnail idea.
    
    

    You write a prompt.
    ChatGPT gives an answer.
    If you don't like the answer, you manually write another prompt.
    You repeat this process until you're satisfied.



You → Prompt
      ↓
ChatGPT → Response
      ↓
Not satisfied?
      ↓
You write another prompt
      ↓
Repeat...



Now imagine repeating this manually 150 times a year.
That's where Loop Engineering shines.



->> What Loop Engineering Does

    Loop eng is not about , finding 1 magical prompt !
    Its about : building a good loop around the LLM !
    & Good Loop needs 3 things :

    I )   Goal : What u want your AI to do ! eg (make the test pass)
    II )  Action ( AI will act as per prompt)
    III ) Stop : U have to tell CLAUDE/CHATGPT when to stop  
                 else Your CLAUDE/CHATGPT keeps on running & token usage will be very high !
                 ie , like U can stop the loop , the moment all the test cases pass or u can stop it after 15 runs !


    Infact there are 6 major components for Loop Engineering which we will discuss later in this coursw !



    First u have to open CLAUDE CODE or u can run this in CODEX or into CURSOR !
    So now lets design a loop !
    ## first is u have to explain GOAL !
    
    GOAL:
    Create a complete YouTube content package on the given topic.



    ## Second is u have to explain ACTION !
    ACTION:
    Repeat until everything meets the quality standards:

    1. Research the latest information.
    2. Create a content outline.
    3. Write the YouTube script.
    4. Fact-check and verify the content.
    5. Improve storytelling and engagement.
    6. Optimize for YouTube SEO.
    7. Generate SEO-friendly YouTube titles.
    8. Generate thumbnail ideas.



    ## Then u have to add STOP condition for your Agent !
    Stop when:
    - The research is accurate and up-to-date.
    - The script is engaging and easy to understand.
    - All facts are verified.
    - SEO is optimized.
    - Titles and thumbnail ideas are ready.
    - OR after 15 iterations
        The AI keeps refining its own work without waiting for another human prompt.






    Que : Now the que is : I can set this task on schedule
                Search today's AI news.
                Write a script.
                Generate a title.
                Create a thumbnail idea.
    
        then why Loop enginerring needed ?




Scheduled task answers : "When should Agent run?"

Loop Engineering answers:
 "What should the AI keep doing, checking, verifying, remembering, and improving while it runs?"











    u don't have to build these loops by hand :
    we have 2 things :

    Lang-graph helps us to define loops as a set of steps
    CLAUDE Agent SDK : Agent SDK runs whole cycle for you







'''
```
