```python
'''

Hi , In this module , we will cover 6 key components to design any Loop
    So Lets consider a real-world scenario to understand : 6 key components 
    that allow AI/Agents to work continuously with minimal human involvement




Example: AI Job Search Loop


Instead of manually visiting LinkedIn, Naukri, Indeed & 100s of sites etc. every day, 
you create an AI system/LOOP that does the work on your behalf & select Best Jobs as per my Profile !
                    OR
we will design a self system in a such a way that work will be done autonomously without promting !




    GOAL: Find me a good job that matches my "Content Strategist" profile.

    1. AUTOMATION
        Run every day at 9:00 AM.
        The AI automatically starts searching without you having to open any job portal.

    2. RULES: Marketing/Content roles only, Delhi NCR, 8 LPA+ (Lakhs Per Annum)
    
    3. ACTION: Customize resume and apply

    4. VERIFIER: Duplicate check, red-flag job description check

    5. MEMORY: In the sheet:
    
    6. Stop : Stop after receiving 3 interview call/15 runs









1.. Automation
    The first component of Loop Engineering is Automation.

    Automation answers one simple question:

    "When should the AI start working?"

     Here you tell the AI : Every day at 9:00 AM, check all job portals for new jobs.

    Instead of you opening LinkedIn, Naukri, or Indeed every morning, the AI agent does it automatically.
    So the loop starts like this:

    Time: 9:00 AM
        │
        ▼
    Automation triggers the Agent
        │
        ▼
    Agent opens LinkedIn, Naukri, Indeed, etc.
        │
        ▼
    Searches for new jobs matching your criteria







Open CLAUDE desktop->> Go to CLAUDE code section
/loop GOAL: Find me a good job that matches my "Content Strategist" profile.

AUTOMATION:
Check job sites every morning at 9 AM (LinkedIn, Naukri, Indeed).


'''
```

# 2.. Skills , Plugins & Connectors

```python
'''

2.. Skills , Plugins & Connectors

    The second component of Loop Engineering is Rules/Skills.

    Rules tell the AI what to look for and what to ignore.
    Without rules, the AI might search for any job, such as:

    Teacher
    Software Engineer
    Sales Executive
    Accountant
    Marketing Manager

    That's not what you want.

    So you define clear rules (filters).

    Example Rules
        Role: Marketing / Content only
        Location: Delhi NCR
        Salary: 8 LPA or above
        Experience: 2+ years
        Employment Type: Full-time

Now, whenever the AI searches job portals, it will:

✅ Consider jobs that match these rules.



    Similarly if u want your Agent to fetch Jobs only from naukri or LInkedIN , u can use plugins & connector
    to first integrate Indeed & Claude !









/loop GOAL: Find me a good job that matches my profile.

AUTOMATION:
Check job sites every morning at 9 AM (LinkedIn, Naukri, Indeed).

RULES (preferences):
• Only look at "Marketing Manager" or "Content Strategist" roles
• Location: Delhi NCR or Remote
• Minimum salary: 8 LPA
• Company size: 50+ employees (startups feel too risky to me)




'''
```

# 4.. Sub-Agent !

```python
'''

4.. Verifier Agent
    The fourth component of Loop Engineering is the Verifier..
    In market,  there are fake/scam jobs , 
    there are some duplicate jobs as well
    

    So A Verifier is a separate AI sub-agent whose only job is to review and validate the work done by the main agent.

        Is this a real job or a fake/scam posting?
        Is this a duplicate job that we've already applied to !
        
        
    ie , Only verified jobs go into the spreadsheet.
        
        
        
        
 /loop GOAL: Find me a good job that matches my profile.

AUTOMATION:
Check job sites every morning at 9 AM (LinkedIn, Naukri, Indeed).

RULES (preferences):
• Only look at "Marketing Manager" or "Content Strategist" roles
• Location: Delhi NCR or Remote
• Minimum salary: 8 LPA
• Company size: 50+ employees (startups feel too risky to me)

ACTION:
For every job that matches, create a spreadsheet of the jobs that matches with my profile.

VERIFIER (double-check before applying):
• Does the job description contain any red flags (e.g. "unpaid", "no fixed salary")?       
       

















Imagine YOu agents are doing heavy work & they are searching across 1000's sites , 
you can split the work across multiple specialized agents.

For our job-search example:

                    MAIN AGENT
                  "Find me a good job"
                         │
          ┌──────────────┼──────────────┐
          ↓              ↓              ↓
   LINKEDIN AGENT   NAUKRI AGENT   INDEED AGENT
          │              │              │
     Search jobs     Search jobs    Search jobs
     Apply rules     Apply rules    Apply rules
          │              │              │
          └──────────────┼──────────────┘
                         ↓
                    MAIN AGENT
                         │
                  Combine results
                         ↓
                     VERIFY
                         ↓
                  Save to Sheet








'''
```

# 5.. Memory !

```python
'''



5.. Memory
    The fifth component of Loop Engineering is Memory.

    Memory keeps a record of everything the AI has done.

    here , after applying to jobs, the AI saves all the details in a spreadsheet, database, or Markdown file.
    It records things like:

    Which companies were found today.
    Which jobs were applied to.
    Which applications are still pending.
    Which companies have responded.
    
    Example : Date: 07 Aug

    ✓ Google - Content Strategist - Applied
    ✓ Adobe - Marketing Manager - Applied
    ⏳ HubSpot - Pending
    ❌ Amazon - Already Applied




    The next day, before searching again, the AI first reads this memory..
    
    
    
    




/loop GOAL: Find me a good job that matches my profile.

AUTOMATION:
Check job sites every morning at 9 AM (LinkedIn, Naukri, Indeed).

RULES (preferences):
• Only look at "Marketing Manager" or "Content Strategist" roles
• Location: Delhi NCR or Remote
• Minimum salary: 8 LPA
• Company size: 50+ employees (startups feel too risky to me)

ACTION:
For every job that matches, create a spreadsheet of the jobs that matches with my profile.

VERIFIER (double-check before applying):
• Does the job description contain any red flags (e.g. "unpaid", "no fixed salary")?

MEMORY / LOG:
Record every applied job in a tracking sheet:
"[date] Company X - Role Y - Status: Applied"

This prevents applying to the same job twice and keeps a record of how many places I've applied to.


'''
```

# 6.. Stop

```python
'''


    
6.. Stop Condition (Simple & Layman)

    The sixth and final component of Loop Engineering is the Stop Condition..
    A Stop Condition tells the Agent when to stop running the loop.

    Without a stop condition, the AI would keep searching and applying for jobs forever.

    Example


    You can define the stop condition as:

    ✅ Stop after receiving 3 interview calls, OR
    ✅ Stop after 30 days (1 month), even if 3 interviews haven't been received.
















    /loop GOAL: Find me a good job that matches my Marketing Manager profile..

AUTOMATION:
Check job sites every morning at 9 AM (LinkedIn, Naukri, Indeed).

RULES (preferences):
• Only look at "Marketing Manager" or "Content Strategist" roles
• Location: Delhi NCR or Remote
• Minimum salary: 8 LPA
• Company size: 50+ employees (startups feel too risky to me)

ACTION:
For every job that matches, create a spreadsheet of the jobs that matches with my profile.

VERIFIER (double-check before applying):
• Does the job description contain any red flags (e.g. "unpaid", "no fixed salary")?

MEMORY / LOG:
Record every applied job in a tracking sheet:
"[date] Company X - Role Y - Status: Applied"

This prevents applying to the same job twice and keeps a record of how many places I've applied to.

STOP CONDITION:
Stop and give me a summary once either:
• The spreadsheet contains a record of 20 jobs.







'''
```
