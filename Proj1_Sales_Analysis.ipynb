````python
'''


Data Analytics Workflow :

Data Collection ->> Load data ->> Clean data ->> Analyse data ->> Create dashboards !








With traditional Prompt Engineering, we often work step-by-step:



    First we give prompt to collect data
    First we give prompt to read data
    Then we ask CLAUDE to check data ( anamolies , missing data types , etc)
    Then we ask CLAUDE to clean it ( fix missing values , )
    Then we ask Claude to analyse it 
    Then we ask Claude to create Dashboard
    Review the output.
    
    
    
    
    If something goes wrong...
    ↓
    Refine the prompt
    ↓
    Run again
    ↓
    Refine again
    ↓
    Run again
    ↓
    ...


'''
````

````python
'''

Instead of repeatedly driving the workflow with prompts, we design the workflow once.




The Actual loop should look like this : 


             ┌──────────────────────┐
             │      USER REQUEST     │
             │ "Build sales dashboard"│
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │       AUTOMATION      │
             │    Start the loop     │
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │        ACTION         │
             │ Fetch / clean / build │
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │ QA Validator SUB-AGENT  │
             │  Validate the output │
             └──────────┬───────────┘
                        ↓
             ┌──────────────────────┐
             │        MEMORY         │
             │ Record result/failure │
             └──────────┬───────────┘
                        ↓
                 ┌──────────────┐
                 │     DONE?    │
                 └──────┬───────┘
                    YES │ NO
                        │  └──────────┐
                        ↓             │
                      STOP            │
                                      ↓
                                READ MEMORY
                                      │
                                      └──→ ACTION







Prompt Engineering:
    You guide the AI  by repeatedly giving and refining prompts.

Loop Engineering:
    You design the system once, define the actions, validation, memory, and stop conditions — 
    then trigger the workflow with one command.




Prompt Engineering: Keep prompting until the task is done.
Loop Engineering: Build the loop once, then let the loop keep working until the task is done.








'''
````

````python
'''





# Sales Dashboard Loop

## Project Goal
Build a reliable sales analytics dashboard from the latest sales CSV through an iterative loop..



The loop should:
**Fetch → Clean → Analyze → Visualize → Validate → Fix → Repeat → Stop**



## Project Structure

```text
Sales/
│
├── .claude/
│   ├── agents/
│   │   └── qa-validator.md
│   └── skills/
│       ├── data-cleaner/
│       │   └── SKILL.md
│       └── chart-builder/
│           └── SKILL.md
│
├── data/
│   └── sales_data.csv
│
├── memory/
│   └── state.json
│
├── reports/
│   └── dashboard.md
│
├── scripts/
│   └── build_dashboard.py
│
├── output/
│   └── dashboard.html
│
├── .mcp.json
└── CLAUDE.md
```

### What each part does

- **GitHub MCP / `.mcp.json`** → connects the agent to the GitHub repository where the latest CSV arrives.
- **`data/sales_data.csv`** → local working copy of the GitHub sales data.
- **`data-cleaner` Skill** → defines how the raw sales data is inspected and cleaned.
- **`chart-builder` Skill** → defines the dashboard charts and presentation rules.
- **`build_dashboard.py`** → Action that performs the repeatable data-processing and dashboard-building work.
- **`qa-validator.md`** → Sub-agent that independently checks the dashboard and numbers.
- **`state.json`** → Memory of loop attempts, failures, and fixes.
- **`dashboard.md`** → Human-readable summary of the final run.
- **`dashboard.html`** → Final deliverable for the user.
- **`CLAUDE.md`** → Overall rules, workflow, and stop conditions.







## Workflow

1. Fetch the latest `sales_data.ftr` from the Loop_Engineering GitHub repository using GitHub MCP, then convert it to CSV..
2. Save/use it as the local working copy in `data/sales_data.csv`.
3. Inspect the raw data.
4. Clean invalid, blank, duplicate rows using the `data-cleaner` Skill.
5. Analyze the cleaned sales data.
6. Build the dashboard using the `chart-builder` Skill and `scripts/build_dashboard.py`.
7. Run the `qa-validator` sub-agent.
8. If QA fails, record the problem in `memory/state.json`, fix it, and run the workflow again.
9. If QA passes, update the report and stop.



## Dashboard Requirements


Create these five beginner-friendly KPIs:

1. **Total Revenue** — total sales/revenue generated.
2. **Total Orders** — total number of orders.
3. **Units Sold** — total quantity sold.
4. **Average Order Value** — average revenue per order.
5. **Top-Selling Product** — product with the highest total sales.



Create these 5 charts:

1. **Monthly Sales** — total sales by month.
2. **Top 10 Products** — products ranked by total sales.
3. **Sales by City** — total sales by city.
4. **Sales by Category** — total sales by product category.
5. **Quantity vs Sales** — relationship between quantity sold and sales.

All metrics must be calculated from the cleaned data.



## Definition of DONE

The loop is complete only when:

- Latest sales data was fetched successfully from GitHub.
- The data is readable and cleaned.
- The five KPI's are present.
- The five required charts are present.
- Dashboard calculations are based on cleaned data.
- `output/dashboard.html` exists and opens successfully.
- QA Validator returns **PASS**.
- `memory/state.json` records the final successful attempt.











## Loop Rules

- Never invent sales values, products, dates, or cities.
- Never modify the source CSV in GitHub.
- Use `data/sales_data.csv` only as the local working copy.
- Do not silently discard valid sales records.
- Record important failures and fixes in `memory/state.json`.
- Make one logical fix at a time when QA fails.
- Re-run validation after every fix.
- Do not declare success unless all DONE conditions pass.













'''
````

## 6.. Delegating task to QA Validator Sub-Agent !

````python
'''



[show VS Code]
Hi! In the last session, we defined our Skills.



    [show sales workflow]
    Our chart-builder Skill tells Claude Code how the Sales Dashboard should be built.

    For example, it defines:

    Which KPIs to calculate
    Which charts to create
    How sales should be calculated
    Where the final dashboard should be saved

But there's an important question:
How do we know that the dashboard Claude created is actually correct?

That's where our QA Validator Sub-Agent comes in.




What does the QA Validator do?
The job of this sub-agent is simple:

QA VALIDATOR
     │
     ├── 🔍 Inspect the generated dashboard
     │
     ├── 🧮 Check KPI calculations
     │
     ├── 📊 Check required charts
     │
     ├── 🔎 Compare results with cleaned data
     │
     └── ✅ PASS / ❌ FAIL







Skill tells the main agent HOW to build the dashboard.
QA Sub-Agent tells the main agent WHETHER the result is correct.
Main Agent performs the work and fixes any problems.


And here's the important part

    The QA Validator does NOT fix the dashboard.
    It only:
    Checks the result and reports what is wrong.

    The main Claude Code agent is responsible for deciding what needs to be fixed and applying the fix.






'''
````

````python
'''




---
name: qa-validator
description: Independently validate the sales dashboard and its calculations.
tools: Read, Grep, Glob
---

# Sales Dashboard QA Validator

You are a read-only QA sub-agent.

## Check

- `data/sales_data.csv` is readable.
- Required columns are present.
- Blank/repeated-header rows were handled.
- Numeric fields are valid.
- Sales calculations use `Quantity Ordered × Price Each`.
- The dashboard contains all five required KPIs:
  - Total Revenue
  - Total Orders
  - Units Sold
  - Average Order Value
  - Top-Selling Product
- The dashboard contains all five required charts:
  - Monthly Sales
  - Top 10 Products
  - Sales by City
  - Sales by Category
  - Quantity vs Sales
- Dashboard values are supported by the cleaned data.
- `output/dashboard.html` exists.
- The dashboard is not empty or obviously broken.

## Rules

- Do not modify data or project files.
- Do not guess missing values.
- Base findings on evidence from the current project.

## Return

Return exactly one final status:

**PASS** — all checks succeed.

**FAIL** — include the specific problem(s) that must be fixed.









'''
````

## 7.. Add memory to Agent

````python
'''


Memory
  ↓
is like a Notebook which tells Agent
"Here's what we've done so far" eg (so it tells goal, previous attempts, results, and current status.)





iteration
→ 🔄 Which attempt we're currently on.

max_iterations
→ 🔢 Maximum number of attempts allowed.


clean_rows
→ 🧹 Number of valid records remaining after cleaning.

qa_status
→ ✅ Latest validation result from the QA sub-agent.

last_issue
→ ❌ What went wrong during the previous attempt.

last_fix
→ 🔧 What the agent changed to fix the problem.

kpis
→ 📈 The important business metrics produced by the successful run.

dashboard
→ 📊 Tracks whether the required KPIs and charts were actually created.










How memory Works :


Iteration 1
     ↓
Dashboard created
     ↓
QA Sub-Agent FAIL
     ↓
Memory updated
     │
     ├── What failed?
     └── What needs fixing?
     ↓
Iteration 2
     ↓
Main Agent reads Memory
     ↓
Knows previous failure
     ↓
Makes targeted fix
     ↓
QA Sub-agent PASS
     ↓
Memory updated
     ↓
Loop STOP



Memory is the loop's notebook.
It records what happened in previous attempts so the agent doesn't have to start from zero every time.


'''
````
